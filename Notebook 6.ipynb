## Sample DataFrame

```python
import pandas as pd

data = {
    "Name": ["Asha", "Ravi", "Kiran", "Priya", "Arun", "Meena", "Ravi", "Asha"],
    "Age": [21, 22, 20, 23, 21, 22, 22, 21],
    "Marks": [82, 76, 91, 88, 69, 95, 76, 84],
    "City": ["Bengaluru", "Mysuru", "Hassan", "Bengaluru", "Hassan", "Mysuru", "Mysuru", "Bengaluru"]
}

df = pd.DataFrame(data)
df
```

## 1. head()

### Definition
The head() method displays the first rows of a Pandas DataFrame. By default, it shows the first 5 rows, but we can specify another number. It is commonly used after loading data to quickly inspect the beginning of a dataset.

### Real-Life Example
A company may have thousands of customer records. An analyst can use head() to check the first few records and confirm that the columns and values were loaded correctly. This gives a quick initial view without opening the complete dataset.

### AI/ML Example
In AI/ML, head() is commonly used before preprocessing and model training. It gives a quick view of features, values, and the overall structure of the dataset. This helps us understand what kind of information is available for the model.

### Explanation
df.head() displays the first 5 rows by default. We can use df.head(3) to display only the first 3 rows. It only previews the data and does not modify the original DataFrame.

### Python Implementation


```python
print(df.head())
print(df.head(3))
```

## 2. tail()

### Definition
The tail() method displays the last rows of a Pandas DataFrame. By default, it shows the last 5 rows, but we can specify another number. It is useful for checking the ending records of a dataset.

### Real-Life Example
A company may continuously add new sales records at the bottom of a dataset. An analyst can use tail() to inspect the latest entries and verify that new records were added correctly. This is useful for regularly updated data.

### AI/ML Example
In AI/ML, tail() helps inspect the final records after loading, sorting, or processing data. It can help confirm that a dataset is complete and transformations were applied correctly. It is especially useful with ordered or time-based data.

### Explanation
df.tail() returns the last 5 rows by default. We can use df.tail(3) to display the last 3 rows. It only displays selected rows and does not change the original DataFrame.

### Python Implementation


```python
print(df.tail())
print(df.tail(3))
```

## 3. sample()

### Definition
The sample() method returns random rows from a Pandas DataFrame. We can specify the number of random records using the n parameter. It helps us inspect different parts of a dataset instead of only the beginning or end.

### Real-Life Example
A company may have thousands of customer feedback records. An analyst can randomly select a few records to review different responses. This provides a more varied view of the dataset.

### AI/ML Example
In AI/ML, sample() can randomly inspect examples before preprocessing or training. Random inspection may reveal unusual values, incorrect labels, or inconsistent records. It is also useful for quick experimentation with a small subset.

### Explanation
df.sample(n=3) randomly selects 3 rows. random_state can be provided when we want the same random sample every time the code runs. The original DataFrame remains unchanged.

### Python Implementation


```python
print(df.sample(n=3, random_state=42))
```

## 4. info()

### Definition
The info() method provides a concise summary of a DataFrame. It shows column names, non-null counts, data types, index information, and approximate memory usage. It is one of the main methods used during initial data inspection.

### Real-Life Example
A company receives a customer dataset from another department. An analyst can use info() to check available columns, their data types, and whether values may be missing. This helps identify basic data-quality issues quickly.

### AI/ML Example
In AI/ML, info() is useful during the initial inspection stage. It helps identify missing values and incorrect data types before preprocessing. For example, a numeric feature stored as object data may need conversion before model training.

### Explanation
df.info() prints a structural summary rather than a normal table. Non-Null Count helps identify possible missing values, while Dtype shows how Pandas stores each column. It also gives an estimate of memory usage.

### Python Implementation


```python
df.info()
```

## 5. describe()

### Definition
The describe() method generates summary statistics for DataFrame columns. For numerical columns, it shows values such as count, mean, standard deviation, minimum, maximum, and quartiles. It provides a quick statistical overview of numerical data.

### Real-Life Example
A company can use describe() on salary or sales data to understand average, minimum, maximum, and spread. This avoids calculating each statistic separately. It is useful during initial business analysis.

### AI/ML Example
In AI/ML, describe() helps us understand numerical feature distributions before preprocessing and training. It can reveal unusual ranges and differences in feature scales. This can guide scaling, outlier detection, and further analysis.

### Explanation
df.describe() calculates descriptive statistics mainly for numerical columns by default. The 25%, 50%, and 75% values represent quartiles, and 50% represents the median. These statistics describe the center, spread, and range of features.

### Python Implementation


```python
print(df.describe())
```

## 6. shape

### Definition
The shape attribute returns the dimensions of a DataFrame as a tuple. The first value is the number of rows and the second is the number of columns. It gives a quick understanding of dataset size.

### Real-Life Example
A company can use shape to check how many customer records and fields are available in a received dataset. This avoids manually counting rows and columns. It also helps confirm that the expected amount of data was loaded.

### AI/ML Example
In AI/ML, shape is commonly checked before and after preprocessing. It shows how many samples and features are available for model development. We can compare shapes after filtering, removing missing values, or selecting features.

### Explanation
df.shape returns a tuple in the form (rows, columns). For example, (8, 4) means 8 records and 4 columns. shape is an attribute, so it is written without parentheses.

### Python Implementation


```python
print(df.shape)
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

## 7. columns

### Definition
The columns attribute returns the column labels of a Pandas DataFrame. It allows us to see all field or feature names available in the dataset. These names are used when selecting, filtering, cleaning, and transforming data.

### Real-Life Example
A company dataset may contain Name, Age, Salary, and Department columns. An analyst can use columns to quickly check whether all expected fields are present. It can also help identify incorrect or inconsistent column names.

### AI/ML Example
In AI/ML, columns helps identify available input features and possible target columns. We inspect column names before feature selection and preprocessing. It is also useful when selecting or dropping specific features.

### Explanation
df.columns returns an Index object containing the DataFrame column names. We can convert it to a normal Python list using list(df.columns). It reports the labels without modifying the dataset.

### Python Implementation


```python
print(df.columns)
print(list(df.columns))
```

## 8. dtypes

### Definition
The dtypes attribute shows the data type of every DataFrame column. Common types include int64, float64, object, bool, and datetime types. It helps us understand how Pandas is storing each feature.

### Real-Life Example
A company may have an Age column that should contain numbers but is stored as text. Checking dtypes helps an analyst identify this problem before calculations. The column can then be converted to the correct data type.

### AI/ML Example
In AI/ML, correct data types are important before preprocessing and training. Numerical features may require scaling, while categorical features may require encoding. dtypes helps us decide which preprocessing operation is suitable for each column.

### Explanation
df.dtypes returns the data type associated with each column. Text columns may appear as object or string types, while integer columns may appear as int64. Incorrect types can later be converted using methods such as astype().

### Python Implementation


```python
print(df.dtypes)
```

## 9. memory_usage()

### Definition
The memory_usage() method shows how much memory is used by each DataFrame column. It can also include memory used by the DataFrame index. This is useful when large datasets consume significant system memory.

### Real-Life Example
A company processing millions of transaction records may need to identify memory-heavy columns. memory_usage() provides this information so inefficient data types or unnecessary columns can be optimized. This can reduce resource usage.

### AI/ML Example
In AI/ML, large datasets can consume significant RAM during preprocessing and training. memory_usage() helps identify memory-heavy features before expensive operations. Optimizing memory can make data pipelines faster and more efficient.

### Explanation
df.memory_usage() reports memory consumption in bytes for each column and the index. deep=True gives a more detailed estimate for object-type columns such as strings. The values can be summed to estimate total DataFrame memory.

### Python Implementation


```python
print(df.memory_usage(deep=True))
print("Total memory:", df.memory_usage(deep=True).sum(), "bytes")
```

## 10. nunique()

### Definition
The nunique() method counts the number of unique values in each DataFrame column. It helps us understand how many different values or categories are present. This is especially useful when inspecting categorical features.

### Real-Life Example
A company may want to know how many cities, departments, or product categories are present in its data. nunique() gives these counts quickly without listing every value. This helps summarize the variety in each field.

### AI/ML Example
In AI/ML, nunique() helps measure the cardinality of features before preprocessing. A categorical feature with many unique values may require special encoding or feature engineering. It can also help identify identifier-like columns.

### Explanation
df.nunique() counts distinct non-null values in every column by default. It can also be applied to one column such as df['City'].nunique(). It helps us understand feature diversity without changing the DataFrame.

### Python Implementation


```python
print(df.nunique())
print("Unique cities:", df["City"].nunique())
```

## 11. value_counts()

### Definition
The value_counts() method counts how many times each unique value appears in a column. It normally returns values from the most frequent to the least frequent. It is especially useful for understanding categorical data distributions.

### Real-Life Example
A company may want to know how many customers belong to each city or product category. value_counts() gives the frequency of every category directly. This makes it easy to understand the distribution of business data.

### AI/ML Example
In AI/ML, value_counts() is commonly used to inspect class distributions in classification datasets. It can reveal whether one target class has many more samples than another. This helps identify class imbalance before model training.

### Explanation
df['City'].value_counts() counts how many times each city appears. The same method can be applied to a target column to inspect class frequencies. normalize=True can be used to obtain proportions instead of raw counts.

### Python Implementation


```python
print(df["City"].value_counts())
print(df["City"].value_counts(normalize=True))
```
