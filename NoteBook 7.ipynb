# Notebook 7 – Selecting & Filtering Data

## Sample DataFrame

```python
import pandas as pd

data = {
    "Name": ["Asha", "Ravi", "Kiran", "Priya", "Arun", "Meena", "John", "Neha"],
    "Age": [21, 22, 20, 23, 21, 24, 22, 20],
    "Marks": [82, 76, 91, 88, 69, 95, 79, 86],
    "City": ["Bengaluru", "Mysuru", "Hassan", "Bengaluru",
             "Hassan", "Mysuru", "Bengaluru", "Mysuru"]
}
df = pd.DataFrame(data)
df
```

## 1. loc

### Definition
loc selects rows and columns using labels or names. We can select particular rows, ranges, and specific columns. It is useful when we know the index labels or column names.

### Real-Life Example
A company analyst can select employee names and salaries for required records using their labels. This avoids displaying unnecessary columns. It makes focused analysis easier.

### AI/ML Example
In AI/ML, loc helps select required features or labeled records before preprocessing. It can also update selected values. This helps prepare relevant data for analysis and training.

### Explanation
df.loc[] mainly works with labels. The part before the comma selects rows and the part after the comma selects columns. A colon can select all rows or a range of labels.

### Python Implementation


```python
print(df.loc[0:2, ["Name", "Marks"]])
print(df.loc[:, ["Name", "City"]])
```

## 2. iloc

### Definition
iloc selects rows and columns using integer positions instead of labels. Position counting starts from 0 like Python lists. It is useful when we know where the required data is located.

### Real-Life Example
An analyst may want the first three employee records and first two columns. iloc selects them using numerical positions. The actual labels or column names are not required.

### AI/ML Example
In AI/ML, iloc can select input features and target columns by position. It can also create small row subsets for experiments. This is useful when the dataset structure is known by position.

### Explanation
df.iloc[ ] uses integer positions for rows and columns. Normal Python slicing rules apply, so the ending position is excluded. For example, 0:3 selects positions 0, 1, and 2.

### Python Implementation


```python
print(df.iloc[0:3, 0:2])
print(df.iloc[:4])
```

## 3. Boolean Indexing

### Definition
Boolean Indexing selects rows using conditions that produce True or False. Rows with True are returned while False rows are excluded. It is one of the most common Pandas filtering techniques.

### Real-Life Example
A college can find students who scored more than 80 marks. A Boolean condition checks every Marks value and returns only matching students. This quickly identifies high-performing students.

### AI/ML Example
In AI/ML, Boolean Indexing helps remove invalid records or select samples meeting specific conditions. We may select customers above a certain age or records with valid values. It is common during cleaning and preprocessing.

### Explanation
The expression df['Marks'] > 80 creates a Boolean Series. Passing that condition inside df[] returns only rows where the condition is True. The original DataFrame is not changed.

### Python Implementation


```python
condition = df["Marks"] > 80
print(condition)
print(df[condition])
```

## 4. Filtering Rows

### Definition
Filtering rows means selecting records that satisfy one or more conditions. Pandas allows conditions to be combined using &, |, and ~. This creates a smaller dataset containing only relevant records.

### Real-Life Example
A company may want customers from Bengaluru whose purchase amount is above a limit. Multiple conditions can be combined to return only those customers. This helps focus on a specific group.

### AI/ML Example
In AI/ML, filtering removes unwanted observations and creates specific subsets. We can keep records within valid ranges or selected categories. Clean and relevant records improve preprocessing quality.

### Explanation
Each condition should normally be inside parentheses when conditions are combined. & means AND, while | means OR. Pandas checks each row and returns records satisfying the complete condition.

### Python Implementation


```python
filtered_df = df[(df["City"] == "Bengaluru") & (df["Marks"] > 80)]
print(filtered_df)
```

## 5. query()

### Definition
query() filters DataFrame rows using a readable condition written as an expression. Column names can usually be written directly in the expression. It can make multi-condition filtering easier to read.

### Real-Life Example
A company can find employees above a salary limit and from a particular department using query(). Both conditions can be written in one readable statement. This makes filtering code easier to understand.

### AI/ML Example
In AI/ML, query() can create dataset subsets based on feature conditions. It is useful during exploratory analysis, cleaning, and preprocessing. Its readable syntax is convenient when several conditions are involved.

### Explanation
df.query() accepts a filtering expression as a string. Pandas evaluates the expression using DataFrame columns and returns matching rows. Comparison and logical operators can be used inside the query.

### Python Implementation


```python
print(df.query("Marks > 80"))
print(df.query("Marks > 80 and Age >= 21"))
```

## 6. isin()

### Definition
isin() checks whether each value belongs to a given collection of values. It returns True for matches and False for non-matches. It is useful when filtering several possible categories at once.

### Real-Life Example
A company may want customers living in Bengaluru or Mysuru. isin() checks both cities together instead of requiring separate equality conditions. This keeps the filtering code simple.

### AI/ML Example
In AI/ML, isin() can select specific categories, classes, or labels from a dataset. We may keep only selected product types or target classes. It is also useful when cleaning categorical features.

### Explanation
df['City'].isin([...]) checks every City value against the supplied list. The resulting Boolean condition can filter the DataFrame. More categories can easily be added to the list.

### Python Implementation


```python
cities = ["Bengaluru", "Mysuru"]
print(df[df["City"].isin(cities)])
```

## 7. between()

### Definition
between() checks whether values fall within a specified lower and upper range. It returns a Boolean result for every value. It provides a simple way to filter numerical or ordered data by range.

### Real-Life Example
A company can identify employees whose salaries fall between two values. between() selects records within that range using one simple condition. This is easier than writing separate lower and upper comparisons.

### AI/ML Example
In AI/ML, between() can keep observations within valid numerical ranges. For example, we may keep ages from 18 to 60 or acceptable sensor values. This helps remove unrealistic records during preprocessing.

### Explanation
df['Marks'].between(75, 90) checks whether each value is within the range. By default, both boundary values are included. The Boolean result is then used to filter the DataFrame.

### Python Implementation


```python
print(df[df["Marks"].between(75, 90)])
```

## 8. Sorting Data

### Definition
Sorting Data means arranging DataFrame records in ascending or descending order based on one or more columns. Pandas provides sort_values() for this purpose. Sorting makes patterns and comparisons easier to inspect.

### Real-Life Example
A company can sort sales from highest to lowest revenue to find top-performing products. Employees can similarly be sorted by salary, age, or joining date. Organized data makes reports easier to understand.

### AI/ML Example
In AI/ML, sorting helps during exploratory analysis and inspection of extreme values. Prediction probabilities can be sorted to view the most confident predictions first. It is also useful for ordered or time-based datasets.

### Explanation
df.sort_values() arranges rows according to selected columns. ascending=True sorts low to high and ascending=False sorts high to low. Multiple columns can also be supplied for multi-level sorting.

### Python Implementation


```python
print(df.sort_values(by="Marks", ascending=False))
print(df.sort_values(by=["City", "Marks"], ascending=[True, False]))
```
