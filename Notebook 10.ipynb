# Notebook 10 – Combining Datasets


## 1. concat()

### Definition
concat() combines multiple Pandas DataFrames along rows or columns. It is useful when datasets have similar structures or need to be placed side by side. The axis parameter controls the direction of combination.

### Real-Life Example
A company receives separate monthly sales reports with the same columns. It can stack those reports into one yearly dataset. This avoids manually copying records from different files.

### AI/ML Example
In AI/ML, training data may arrive in separate batches or files. concat() can combine those batches before cleaning and feature preparation. It can also place feature tables side by side when their indexes align.

### Explanation
pd.concat() accepts a list of DataFrames. axis=0 stacks rows, while axis=1 combines columns by index. ignore_index=True creates a fresh continuous row index when stacking rows.

### Python Implementation


```python
import pandas as pd

jan = pd.DataFrame({'Name': ['Asha', 'Ravi'], 'Marks': [82, 76]})
feb = pd.DataFrame({'Name': ['Kiran', 'Priya'], 'Marks': [91, 88]})

combined = pd.concat([jan, feb], ignore_index=True)
print(combined)
```

## 2. merge()

### Definition
merge() combines DataFrames using matching values from one or more key columns. It works similarly to SQL joins and supports inner, left, right, and outer joins. It is helpful when related information is stored in separate tables.

### Real-Life Example
A company keeps employee names in one table and salaries in another. Both tables contain EmployeeID, so merge() can combine their information. This creates one useful report without duplicating the original data storage.

### AI/ML Example
In AI/ML, features and labels may be stored in different tables. merge() can connect them using a common identifier before model training. Selecting the correct join type helps avoid unexpectedly missing or duplicated records.

### Explanation
pd.merge() matches rows using the on parameter or left_on and right_on. An inner join keeps only matching keys, while a left join preserves every key from the left DataFrame. The validate option can help detect unexpected many-to-many matches.

### Python Implementation


```python
import pandas as pd

students = pd.DataFrame({'StudentID': [1, 2, 3], 'Name': ['Asha', 'Ravi', 'Kiran']})
scores = pd.DataFrame({'StudentID': [1, 2, 3], 'Marks': [82, 76, 91]})

merged = pd.merge(students, scores, on='StudentID', how='inner', validate='one_to_one')
print(merged)
```

## 3. join()

### Definition
join() combines DataFrames primarily using their index labels. It can also join a column in the left DataFrame to the index of the right DataFrame. This is convenient when datasets are already indexed by a common identifier.

### Real-Life Example
A school keeps student contact information and attendance in separate tables indexed by StudentID. join() can place attendance alongside the contact information. The shared index makes the combination simple.

### AI/ML Example
In AI/ML, separate feature sets may share the same sample index. join() can combine these features before preprocessing and model training. Correct index alignment is important so each feature belongs to the right sample.

### Explanation
DataFrame.join() matches index labels by default. The how parameter controls whether matching, left, right, or all indexes are kept. Unlike merge(), it is especially convenient for index-based combinations.

### Python Implementation


```python
import pandas as pd

names = pd.DataFrame({'Name': ['Asha', 'Ravi', 'Kiran']}, index=[101, 102, 103])
marks = pd.DataFrame({'Marks': [82, 76, 91]}, index=[101, 102, 103])

joined = names.join(marks, how='left')
print(joined)
```

## 4. append() – Deprecation and Alternative

### Definition
DataFrame.append() was an older method used to add rows from another DataFrame. It was deprecated in Pandas 1.4 and removed in Pandas 2.0. The recommended alternative is pd.concat(), which supports combining multiple DataFrames.

### Real-Life Example
A company may want to add a new batch of sales records to an existing report. Older scripts used append() for this task. Modern scripts use pd.concat() to perform the same row combination.

### AI/ML Example
In AI/ML, datasets are often expanded with additional training samples. pd.concat() is the supported way to combine batches of records. It also avoids errors caused by calling the removed append() method in newer Pandas versions.

### Explanation
Do not run df.append() in current Pandas because it may raise AttributeError. Use pd.concat([df1, df2], ignore_index=True) instead. Collecting DataFrames and concatenating once is generally preferable to repeatedly adding rows in a loop.

### Python Implementation


```python
import pandas as pd

old_data = pd.DataFrame({'Name': ['Asha'], 'Marks': [82]})
new_data = pd.DataFrame({'Name': ['Ravi'], 'Marks': [76]})

# Alternative to the removed DataFrame.append()
updated = pd.concat([old_data, new_data], ignore_index=True)
print(updated)
```

## 5. combine_first()

### Definition
combine_first() fills missing values in one DataFrame using values from another DataFrame. It aligns rows and columns by their index and column labels. Existing non-missing values in the first DataFrame are preserved.

### Real-Life Example
A company has an incomplete customer contact list and an older backup containing some missing phone numbers. combine_first() can fill gaps from the backup without overwriting existing contact information. This improves completeness while preserving preferred values.

### AI/ML Example
In AI/ML, missing feature values may be available in another trusted dataset. combine_first() can fill these gaps before preprocessing or model training. Alignment by labels helps ensure that replacement values belong to the correct records.

### Explanation
df1.combine_first(df2) uses values from df2 only where df1 has missing values. It aligns both DataFrames by row and column labels and can include labels present in either DataFrame. Non-missing values in df1 take priority.

### Python Implementation


```python
import pandas as pd

primary = pd.DataFrame({'Marks': [82, None, 91]}, index=['Asha', 'Ravi', 'Kiran'])
backup = pd.DataFrame({'Marks': [80, 76, 89]}, index=['Asha', 'Ravi', 'Kiran'])

completed = primary.combine_first(backup)
print(completed)
```
