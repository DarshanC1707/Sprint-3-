# Notebook 12 – String Operations

## Sample DataFrame


```python
import pandas as pd

# One sample dataset for all string operations
records = {
    "Name": [" Asha ", " Ravi ", "Kiran", " Priya ", "Arun"],
    "Full_Name": ["Asha Kumar", "Ravi Sharma", "Kiran Rao", "Priya Singh", "Arun Das"],
    "City": ["Bengaluru", "Blr", "MYSURU", "Hassan", "Blr"],
    "Feedback": [
        "Fast delivery for ORD-1024",
        "Delayed DELIVERY for ORD-2035",
        "Good service",
        "Need refund for ORD-3456",
        "Delivery was on time"
    ]
}

df = pd.DataFrame(records)
print(df)
```

## 1. lower()

### Definition
The lower() string method converts uppercase letters into lowercase letters. In Pandas, we use the .str.lower() accessor to apply it to an entire text column. It is useful for making text consistent before comparison or analysis.

### Real-Life Example
A company may receive customer city names written as Bengaluru, BENGALURU, and bengaluru. Converting them to lowercase makes all three values consistent. This prevents the same city from appearing as separate categories in reports.

### AI/ML Example
In AI/ML, text normalization is often performed before feature extraction or text classification. Lowercasing can reduce differences caused only by capitalization. For example, review words such as GOOD and good can be treated consistently when capitalization is not meaningful.

### Explanation
df["City"].str.lower() returns a new Series containing lowercase strings. The original column remains unchanged unless the result is assigned back. Missing values are generally preserved when using Pandas string methods.

### Python Implementation

```python
print(df["City"].str.lower())
```

## 2. upper()

### Definition
The upper() method converts lowercase letters into uppercase letters. In Pandas, .str.upper() applies the conversion to all string values in a Series. It is helpful when we want standardized codes, labels, or formatted text.

### Real-Life Example
A company may store product codes in mixed case, such as ab12 and AB12. Converting codes to uppercase gives them a common format. This makes matching and reporting easier.

### AI/ML Example
In AI/ML, uppercase conversion may be useful for standardizing categorical codes before merging datasets. For example, department codes coming from different sources can be converted to one format. This reduces accidental mismatches during preprocessing.

### Explanation
df["City"].str.upper() converts text values into uppercase strings. It returns a new Series without changing the original DataFrame. We can store the result in a new column when both original and standardized values are needed.

### Python Implementation

```python
print(df["City"].str.upper())
```

## 3. strip()

### Definition
The strip() method removes extra whitespace from the beginning and end of text. In Pandas, .str.strip() cleans an entire string column at once. It is useful because unwanted spaces can cause values that look identical to be treated as different.

### Real-Life Example
A company might receive customer names with spaces before or after them. For example, " Ravi " and "Ravi" may be treated as different values. Removing the extra spaces makes customer records more consistent.

### AI/ML Example
In AI/ML, extra spaces can create unnecessary categories and mismatched labels. Cleaning them before encoding or joining datasets improves data quality. This is especially useful for categorical features collected from forms or CSV files.

### Explanation
df["Name"].str.strip() removes whitespace from both ends of every text value. It does not remove spaces inside a name such as "Asha Kumar". The cleaned result can be assigned to a column for later processing.

### Python Implementation

```python
print(df["Name"].str.strip())
```

## 4. split()

### Definition
The split() method divides a string into smaller parts using a separator. Pandas provides .str.split() to apply this operation to a text column. It can separate names, words, tags, or other information stored together.

### Real-Life Example
A company may store full names in one column and need separate first and last names. Splitting the name at a space creates individual parts. These parts can then be used for reports or customer communication.

### AI/ML Example
In AI/ML, splitting text is a basic step in many text-processing workflows. For example, a simple sentence can be separated into words before analyzing word frequency. More advanced NLP systems use specialized tokenizers when needed.

### Explanation
df["Full_Name"].str.split(" ") divides each value at spaces and returns a list of parts. Setting expand=True places the parts into separate DataFrame columns. The chosen separator determines how the text is divided.

### Python Implementation

```python
print(df["Full_Name"].str.split(" "))

# Separate first and last names
print(df["Full_Name"].str.split(" ", n=1, expand=True))
```

## 5. contains()

### Definition
The contains() operation checks whether a string includes a particular word or pattern. In Pandas, .str.contains() returns True or False for each row. It is useful for finding matching records within text columns.

### Real-Life Example
A business may want to identify feedback messages mentioning delivery. Using contains() can locate those messages without reading every comment manually. This makes it easier to review a specific type of customer issue.

### AI/ML Example
In AI/ML, contains() can help create simple text-based features or filter records by keywords. For example, messages mentioning "refund" can be selected for further analysis. It is a basic technique and does not replace a full NLP model.

### Explanation
df["Feedback"].str.contains("delivery", case=False, na=False) checks each feedback value. case=False ignores capitalization and na=False treats missing values as nonmatches. regex=False can be used when searching for literal text instead of a regex pattern.

### Python Implementation

```python
mask = df["Feedback"].str.contains("delivery", case=False, na=False, regex=False)
print(df[mask])
```

## 6. replace()

### Definition
The replace() method substitutes one part of a string with another. In Pandas, .str.replace() can update matching text across an entire column. It is useful for correcting repeated spelling, formatting, or naming differences.

### Real-Life Example
A company may store city names using an abbreviation such as "Blr". Replacing the abbreviation with "Bengaluru" gives the records a consistent format. This reduces confusion when grouping or reporting the data.

### AI/ML Example
In AI/ML, string replacement can standardize categories and clean text before preprocessing. For example, inconsistent location abbreviations can be converted to full names. Consistent text values help prevent duplicate categories during encoding.

### Explanation
df["City"].str.replace("Blr", "Bengaluru", regex=False) substitutes the matching text. regex=False treats the search value as ordinary text rather than a regular expression. The operation returns a new Series and does not change the original values unless reassigned.

### Python Implementation

```python
print(df["City"].str.replace("Blr", "Bengaluru", regex=False))
```

## 7. Regular Expressions (Regex)

### Definition
Regular Expressions, or Regex, are patterns used to search, match, and extract text. Pandas string methods can use Regex patterns for flexible text processing. Regex is helpful when text follows a recognizable structure such as digits, email addresses, or codes.

### Real-Life Example
A company may receive customer messages containing order numbers such as ORD-1024. Regex can extract these codes automatically instead of checking every message manually. This helps organize information stored inside unstructured text.

### AI/ML Example
In AI/ML, Regex is commonly used to clean raw text and extract structured features. For example, numeric order IDs can be extracted from support messages before analysis. Regex is useful for pattern-based tasks, although it cannot understand all meanings in natural language.

### Explanation
The pattern r"ORD-\d+" matches the letters ORD-, followed by one or more digits. The \d symbol means a digit, and + means one or more repetitions. .str.extract() returns the first captured match for each row, while unmatched rows receive missing values.

### Python Implementation

```python
print(df["Feedback"].str.extract(r"(ORD-\d+)", expand=False))
```
