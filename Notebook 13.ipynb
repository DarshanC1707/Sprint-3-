# Notebook 13 – Missing Values

## Sample Dataset

The following DataFrame is created in memory. It intentionally contains missing values, so no external CSV or database files are required. Run this cell first.


```python
import pandas as pd

data = {
    "Name": ["Asha", "Ravi", "Kiran", "Priya", "Arun", "Meena"],
    "Marks": [82.0, None, 91.0, 88.0, None, 95.0],
    "City": ["Bengaluru", "Mysuru", None, "Hassan", "Mysuru", None]
}

df = pd.DataFrame(data)
print(df)
```

## 1. isnull()

### Definition
The isnull() method identifies missing values in a Pandas Series or DataFrame. It returns True wherever a value is missing and False where a value is present. Missing values can be represented by NaN, None, or NaT depending on the data type. This is a useful first step before cleaning a dataset.

### Real-Life Example
A college may receive student records where some marks or attendance values are not entered. The analyst can use isnull() to find which records are incomplete. Counting missing values by column helps the college decide which fields need attention.

### AI/ML Example
In AI/ML, missing feature values can prevent models from training correctly or reduce data quality. isnull() helps identify incomplete features before preprocessing. We can measure how much data is missing and then choose a suitable treatment.

### Explanation
df.isnull() creates a Boolean DataFrame showing missing positions. Applying sum() counts the True values in each column because True is treated as 1. This identifies both the location and amount of missing data without changing the original DataFrame.

### Python Implementation


```python
print(df.isnull())
print("Missing values per column:")
print(df.isnull().sum())
```

## 2. notnull()

### Definition
The notnull() method checks whether values are present rather than missing. It returns True for non-missing values and False for missing values. It is the opposite of isnull() and can be used to select valid records. This is useful when we need to work only with available data.

### Real-Life Example
A company may have a customer dataset with some missing email addresses. Using notnull(), the analyst can select customers who have provided an email. The company can then prepare a contact list without including incomplete email records.

### AI/ML Example
In AI/ML, notnull() can select observations where an important feature or target value exists. For example, we may retain only records with known labels for supervised training. This makes it easier to build a valid training dataset.

### Explanation
df.notnull() produces a Boolean result for every cell. We can apply it to a particular column and use the resulting condition to filter rows. This keeps records where the chosen column contains an available value.

### Python Implementation


```python
print(df.notnull())
print("Rows with available Marks:")
print(df[df["Marks"].notnull()])
```

## 3. dropna()

### Definition
The dropna() method removes rows or columns containing missing values. By default, it removes rows with at least one missing value. It also supports options for removing only fully empty rows or targeting selected columns. It should be used carefully because it can discard useful information.

### Real-Life Example
A company may require complete transaction records for a particular report. The analyst can remove incomplete records using dropna() when the missing information cannot be recovered. Before deleting rows, the company should check how many records would be lost.

### AI/ML Example
In AI/ML, dropna() can remove samples with missing values before model training. This may be suitable when only a small number of records are incomplete. However, deleting many rows can reduce training data and introduce bias if the missingness is systematic.

### Explanation
df.dropna() returns a new DataFrame with incomplete rows removed by default. Using subset=["Marks"] removes rows only when Marks is missing. The original DataFrame remains unchanged unless we assign the result back or use inplace=True.

### Python Implementation


```python
print("Drop rows with any missing value:")
print(df.dropna())
print("Drop rows only when Marks is missing:")
print(df.dropna(subset=["Marks"]))
```

## 4. fillna()

### Definition
The fillna() method replaces missing values with specified values. We can fill numerical columns with a mean or median and categorical columns with a suitable label or mode. It helps retain rows instead of removing them. The chosen replacement should make sense for the data.

### Real-Life Example
A school may have a few missing marks in a student report. The analyst could fill missing marks with a reasonable summary value for a preliminary analysis. For a categorical field such as City, a label like Unknown can show that the original value was not available.

### AI/ML Example
In AI/ML, fillna() is a simple way to handle incomplete features during preprocessing. Median imputation can be useful for skewed numerical features, while a separate category can represent missing text values. When training a model, fill values should be learned from the training data only.

### Explanation
fillna() substitutes missing entries with the supplied value. We can calculate the median of the Marks column and use it to fill missing marks. A separate fillna() operation can replace missing City values with Unknown, while leaving the original DataFrame unchanged.

### Python Implementation


```python
filled_df = df.copy()
filled_df["Marks"] = filled_df["Marks"].fillna(filled_df["Marks"].median())
filled_df["City"] = filled_df["City"].fillna("Unknown")
print(filled_df)
```

## 5. interpolate()

### Definition
The interpolate() method estimates missing numerical values from surrounding available values. Linear interpolation fills gaps using values before and after a missing position. It is useful when records have a meaningful sequence, such as measurements over time. It is not appropriate for every dataset or feature.

### Real-Life Example
A company records a machine temperature every hour but misses one reading. Interpolation can estimate the missing temperature from nearby measurements. This can be useful for a continuous sensor series, although the estimated value is not an actual observation.

### AI/ML Example
In AI/ML, interpolation is commonly used when preparing time-series or sensor datasets. It can help preserve a regular sequence of numerical observations before feature engineering. The method should be chosen carefully to avoid using future information in forecasting tasks.

### Explanation
Series.interpolate(method="linear") estimates missing values along the existing order. For example, a gap between 20 and 30 can be filled with 25 when the observations are equally spaced. Interpolation works on numerical values and should only be applied when the sequence has meaningful order.

### Python Implementation


```python
sensor = pd.DataFrame({"Hour": [1, 2, 3, 4, 5], "Temperature": [20.0, 22.0, None, 26.0, 28.0]})
sensor["Temperature_Filled"] = sensor["Temperature"].interpolate(method="linear")
print(sensor)
```

## 6. Missing Value Handling Strategies

### Definition
Missing value handling strategies are methods used to manage incomplete information in a dataset. Common approaches include removing incomplete records, filling values with statistics, using a separate category, and interpolating ordered numerical data. The best approach depends on the feature, amount of missingness, and purpose of analysis.

### Real-Life Example
A company may have missing customer ages, cities, and purchase amounts. It might fill missing ages with a median, use Unknown for missing cities, and review records with missing purchase amounts. Different columns may need different treatments because they represent different types of information.

### AI/ML Example
In AI/ML, missing-value strategies are part of the preprocessing pipeline. We should inspect missingness, choose an appropriate method for each feature, and evaluate how it affects model performance. To avoid data leakage, statistical fill values should be calculated on training data and then applied to validation or test data.

### Explanation
First, count missing values to understand the problem. Next, select a strategy for each column based on its meaning and data type. In this example, we fill numerical Marks with the median and categorical City with Unknown, while retaining the original records.

### Python Implementation


```python
clean_df = df.copy()
print("Before handling missing values:")
print(clean_df.isnull().sum())

clean_df["Marks"] = clean_df["Marks"].fillna(clean_df["Marks"].median())
clean_df["City"] = clean_df["City"].fillna("Unknown")

print("After handling missing values:")
print(clean_df.isnull().sum())
print(clean_df)
```
