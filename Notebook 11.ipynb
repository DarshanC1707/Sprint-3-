## 1. Datetime

### Definition
Datetime represents both a calendar date and a clock time in one value. Pandas supports datetime data for working with dates and times in datasets. It helps organize records that contain events, transactions, or observations.

### Real-Life Example
A company records the exact date and time when each order is placed. Employees use this information to track orders and prepare daily reports. Datetime values make it easier to sort and analyze these events.

### AI/ML Example
In AI/ML, datetime values are common in sales forecasting and user activity datasets. They help arrange observations chronologically before analysis. Models can use features derived from these values to identify time-based patterns.

### Explanation
Python provides datetime objects through the datetime module. Pandas can store datetime values in a Series and work with them efficiently. The following code creates a datetime object and converts it into a Pandas datetime Series.

### Python Implementation


```python
from datetime import datetime
import pandas as pd

order_time = datetime(2026, 10, 8, 14, 30)
print(order_time)

df = pd.DataFrame({"Order_Time": pd.to_datetime(["2026-10-08 14:30", "2026-10-09 09:15"])})
print(df)
```

## 2. Timestamp

### Definition
A Timestamp is Pandas’ representation of a single point in time. It stores a specific date and optionally a time with high precision. It is useful for representing individual time values in data analysis.

### Real-Life Example
A delivery company records when a package reaches a sorting center. Each event can be stored as a Timestamp. These timestamps help the company compare and monitor delivery stages.

### AI/ML Example
In AI/ML, timestamps can identify when sensor readings or transactions occurred. They support chronological sorting and time-based feature engineering. This is important for forecasting and anomaly detection.

### Explanation
pd.Timestamp() creates one Pandas timestamp value. We can access its year, month, day, and other properties. Unlike an entire datetime column, this example represents one specific moment.

### Python Implementation


```python
import pandas as pd

ts = pd.Timestamp("2026-10-08 14:30:00")
print(ts)
print("Year:", ts.year)
print("Month:", ts.month)
print("Day:", ts.day)
```

## 3. Date Parsing

### Definition
Date parsing means converting date information stored as text into actual datetime values. Different datasets may use different date formats. Parsing allows Pandas to perform correct date calculations and sorting.

### Real-Life Example
A company receives an Excel report containing dates written as text. Before calculating monthly sales, the analyst converts the text into proper dates. This prevents incorrect sorting and supports accurate reporting.

### AI/ML Example
In AI/ML, raw datasets often contain date strings that must be converted before preprocessing. Parsing makes it possible to extract time features and order records. This is especially important for time-series models.

### Explanation
pd.to_datetime() converts supported text values into datetime values. The format argument can specify the expected pattern and reduce ambiguity. Once converted, the values can be used for calculations and feature extraction.

### Python Implementation


```python
import pandas as pd

df = pd.DataFrame({"Date_Text": ["08-10-2026", "09-10-2026", "10-10-2026"]})
df["Parsed_Date"] = pd.to_datetime(df["Date_Text"], format="%d-%m-%Y")
print(df)
print(df.dtypes)
```

## 4. Timedelta

### Definition
A Timedelta represents a duration or difference between two dates or times. It can measure intervals such as days, hours, minutes, and seconds. Pandas supports Timedelta operations for duration calculations.

### Real-Life Example
A company wants to calculate how long an order takes from placement to delivery. Subtracting the order date from the delivery date gives the delivery duration. This helps evaluate service performance.

### AI/ML Example
In AI/ML, durations can become useful input features, such as time since last purchase or time between events. These features may help predict customer behavior or equipment failures. Timedelta makes these calculations straightforward.

### Explanation
Subtracting two datetime values produces a Timedelta. We can access the number of days or convert the duration into hours. The example calculates how much time passed between order placement and delivery.

### Python Implementation


```python
import pandas as pd

order_time = pd.Timestamp("2026-10-08 10:00")
delivery_time = pd.Timestamp("2026-10-10 16:00")
duration = delivery_time - order_time
print("Duration:", duration)
print("Days:", duration.days)
print("Hours:", duration.total_seconds() / 3600)
```

## 5. Time Zones

### Definition
Time zones describe the local time offset and regional clock rules associated with a date or time. The same moment can be represented differently in different regions. Pandas supports time-zone-aware datetime values and conversions.

### Real-Life Example
A business with offices in India and the United States schedules an online meeting. Each office sees the meeting in its own local time. Time-zone conversion helps avoid scheduling mistakes.

### AI/ML Example
In AI/ML, global event logs may contain timestamps from multiple regions. Converting them to a common time zone makes chronological comparisons more reliable. It also helps prepare consistent time-based features.

### Explanation
tz_localize() assigns a time zone to a timezone-naive timestamp without changing its displayed clock time. tz_convert() converts an already time-zone-aware timestamp to another zone. The example starts in India and converts the same moment to UTC.

### Python Implementation


```python
import pandas as pd

india_time = pd.Timestamp("2026-10-08 14:30").tz_localize("Asia/Kolkata")
utc_time = india_time.tz_convert("UTC")
print("India:", india_time)
print("UTC:", utc_time)
```

## 6. Extracting Date Features

### Definition
Extracting date features means separating useful parts of a datetime value, such as year, month, day, or weekday. These parts can be added as new DataFrame columns. They help reveal patterns that may not be obvious from a complete date.

### Real-Life Example
A retail company wants to compare sales across months and weekdays. It extracts the month and weekday from each transaction date. This helps identify busy periods and seasonal trends.

### AI/ML Example
In AI/ML, features such as month, weekday, and hour can help models learn seasonal or daily patterns. They are commonly used in demand forecasting and activity prediction. Care is needed to avoid using future information unavailable at prediction time.

### Explanation
The .dt accessor allows Pandas to extract date components from a datetime Series. We create separate columns for year, month, day, and weekday. These features can then be used for grouping, visualization, or model inputs.

### Python Implementation


```python
import pandas as pd

df = pd.DataFrame({"Order_Date": pd.to_datetime(["2026-10-08", "2026-10-09", "2026-11-01"])})
df["Year"] = df["Order_Date"].dt.year
df["Month"] = df["Order_Date"].dt.month
df["Day"] = df["Order_Date"].dt.day
df["Weekday"] = df["Order_Date"].dt.day_name()
print(df)
```
